# Dicionário de Dados

### 0 - brand

**Tipo:** object

**Descrição:** Marca do perfume anunciado no eBay.

**Valores:** Dior, Giorgio Armani, Paco Rabanne, Ralph Lauren, Armaf, entre outras.

### 1 - title

**Tipo:** object

**Descrição:** Título do anúncio do perfume.

**Valores:** Diversos nomes e descrições de perfumes masculinos.

### 2 - type

**Tipo:** object

**Descrição:** Tipo de fragrância do perfume.

**Valores:** Eau de Parfum, Eau de Toilette, Cologne, Perfume.

### 3 - price

**Tipo:** float64

**Descrição:** Preço do perfume anunciado.

**Menor valor:** 3,00

**Maior valor:** 259,09

### 4 - priceWithCurrency

**Tipo:** object

**Descrição:** Preço acompanhado da moeda utilizada no anúncio.

**Valores:** US $24.99, US $59.90, US $120.00.

### 5 - available

**Tipo:** float64

**Descrição:** Quantidade de unidades disponíveis para venda.

**Menor valor:** 2

**Maior valor:** 842

### 6 - availableText

**Tipo:** object

**Descrição:** Informação textual sobre a disponibilidade do produto.

**Valores:** "10 available", "25 available", "100 available".

### 7 - sold

**Tipo:** float64

**Descrição:** Quantidade de unidades vendidas do produto.

**Menor valor:** 1

**Maior valor:** 54.052

### 8 - lastUpdated

**Tipo:** object

**Descrição:** Data e horário da última atualização do anúncio.

**Valores:** Datas e horários variados registrados na plataforma.

### 9 - itemLocation

**Tipo:** object

**Descrição:** Localização do vendedor ou do produto anunciado.

**Valores:** Estados Unidos, Canadá, Reino Unido e outros países.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('tabela.csv', sep = ',')
```

```python
print('Menor preço de perfume: ', df['price'].min())
print('Maior preço de perfume: ', df['price'].max())

print('Menor avaliação: ', df['available'].min())
print('Maior avaliação: ', df['available'].max())

print('Maior venda de perfume: ', df['sold'].max())
print('Menor venda de perfume: ', df['sold'].min())
```

Esse gráfico foi feito para vermos a distribuição entre os preços dos perfumes.



```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df['price'], bins=20)
plt.title('Distribuição dos Preços dos Perfumes')
plt.xlabel('Preço (US$)')
plt.ylabel('Quantidade')
plt.show()
```

```python
vendas_marca = df.groupby('brand')['sold'].sum().sort_values(ascending=False)

print(vendas_marca.head(10))
```

Esse gráfico foi feito para vermos qual perfume foi mais vendido.

```python
import matplotlib.pyplot as plt

vendas_marca.head(10).plot(kind='bar')

plt.title('Top 10 Marcas por Quantidade Vendida')
plt.xlabel('Marca')
plt.ylabel('Quantidade Vendida')
plt.xticks(rotation=45)
plt.show()
```
